```python
# Solução do Desafio: Calculadora de Média

# 1. Entrada de dados
nota_portugues = float(input("Digite a nota de Português (0 a 10): "))
nota_matematica = float(input("Digite a nota de Matemática (0 a 10): "))
nota_redacao = float(input("Digite a nota de Redação (0 a 10): "))

# 2. Cálculo da média
media = (nota_portugues + nota_matematica + nota_redacao) / 3

# 3. Exibição da média e verificação do status
print(f"\nSua média final foi: {media:.2f}")

if media >= 7.0:
    print("Status: Aprovado para a próxima fase!")
elif media >= 5.0:
    print("Status: Em recuperação / Lista de espera.")
else:
    print("Status: Reprovado.")
```
